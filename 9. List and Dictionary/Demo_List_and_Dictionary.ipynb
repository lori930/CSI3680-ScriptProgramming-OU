### 3.2. Conditional update in place
Predict what `alias` will show. `enumerate()` yields `(index, value)` pairs.

`enumerate()` loops over an iterable (like a list, tuple, or string) while keeping track of the index of each element. It returns an enumerate object (i.e., index) and the corresponding element from the iterable. Especially useful when **both the index and value** are needed in a loop.


```python
houses = ["Gryffindor", "Slytherin", "Ravenclaw", "Hufflepuff"]
alias = houses
for i, house in enumerate(houses):
    if house == "Slytherin":
        houses[i] = "Durmstrang"
print("houses:", houses)
print("alias:", alias)
print("Same object:", houses is alias)
```

### 3.3. Build a new list
Predict `houses`, `updated`, and `alias`. Which implementation fits a requirement to preserve the original object?


```python
houses = ["Gryffindor", "Slytherin", "Ravenclaw", "Hufflepuff"]
alias = houses
updated = ["Durmstrang" if house == "Slytherin" else house
           for house in houses]
print("houses:", houses)
print("updated:", updated)
print("alias:", alias)
print("New object:", updated is not houses)
```

**Discuss:** assigning `houses = updated` would rebind the name `houses`. Would that also rebind `alias`?

## 8. Shallow and deep copying
### 8.1. A new outer list
Predict the two contents and the identity comparisons. A shallow copy keeps references to the original elements.


```python
original = [["apple", "banana"]]
copied = original.copy()
copied[0][1] = "cherry"
print("Original:", original)
print("Copy:", copied)
print("Same outer list:", original is copied)
print("Same inner list:", original[0] is copied[0])
```

### 8.2. Independent nested data with deepcopy()
Predict what changes in this ordinary nested list.


```python
import copy
original = [["apple", "banana"]]
copied = copy.deepcopy(original)
copied[0][1] = "cherry"
print("Original:", original)
print("Copy:", copied)
print("Same inner list:", original[0] is copied[0])
```

## 11. One task, three counting patterns
### 11.1. Explicit initialization


```python
message = "banana"
counts = {}
for ch in message:
    if ch not in counts:
        counts[ch] = 0
    counts[ch] += 1
print(counts)
assert counts == {"b": 1, "a": 3, "n": 2}
```

### 11.2. get() with a fallback
Explain how the assignment and the fallback work together. `get()` alone does not insert a key.


```python
message = "banana"
counts = {}
for ch in message:
    counts[ch] = counts.get(ch, 0) + 1
print(counts)
assert counts == {"b": 1, "a": 3, "n": 2}
```

### 11.3. setdefault()
What does this do when the key is already present?


```python
message = "banana"
counts = {}
for ch in message:
    counts.setdefault(ch, 0)
    counts[ch] += 1
print(counts)
assert counts == {"b": 1, "a": 3, "n": 2}
```

## 13. Optional reference and extensions




### 13.1. Eight ways to replace an item
The known-position versions below replace index 1. The conditional versions replace all `Slytherin` values. They match for this particular input, but may differ when duplicates occur or the item moves.

```python
source = ["Gryffindor", "Slytherin", "Ravenclaw", "Hufflepuff"]

houses = source.copy()
houses[1] = "Durmstrang"
print("1. Direct assignment:", houses)

houses = source.copy()
houses[1:2] = ["Durmstrang"]
print("2. Slice assignment:", houses)

houses = source.copy()
for i in range(len(houses)):
    if houses[i] == "Slytherin":
        houses[i] = "Durmstrang"
print("3. Index loop:", houses)

houses = source.copy()
updated = ["Durmstrang" if house == "Slytherin" else house for house in houses]
print("4. Comprehension (new list):", updated)

houses = source.copy()
updated = list(map(lambda house: "Durmstrang" if house == "Slytherin" else house,
                   houses))
print("5. map (new list):", updated)

houses = source.copy()
for i, house in enumerate(houses):
    if house == "Slytherin":
        houses[i] = "Durmstrang"
print("6. enumerate loop:", houses)

houses = source.copy()
houses.pop(1)
houses.insert(1, "Durmstrang")
print("7. pop and insert:", houses)

houses = source.copy()
del houses[1]
houses.insert(1, "Durmstrang")
print("8. del and insert:", houses)
```

### 13.2. Parameter mutation versus rebinding
Predict what the caller observes in each case.


```python
def mutate(values):
    values.append(3)

def rebind(values):
    values = values + [3]

data = [1, 2]
mutate(data)
print("After mutation:", data)

data = [1, 2]
rebind(data)
print("After rebinding:", data)
```

### 13.3. Sorting Hat extension
The function below stores decisions in memory. It is **defined but not called**, so Run all never waits for keyboard input. To try it interactively, run the commented call in a separate cell.


```python
def sorting_hat():
    sortings = {"Harry": "Gryffindor", "Luna": "Ravenclaw"}
    while True:
        name = input("Enter a name (blank to quit): ").strip()
        if not name:
            break
        if name in sortings:
            print(f"{name} is in {sortings[name]}!")
        else:
            house = input("Which house? ").strip()
            sortings[name] = house
            print("Sorting results updated.")
    return sortings

# To run interactively, uncomment:
# saved_sortings = sorting_hat()
```

### 13.4. Membership alternatives
For exact membership, `in` expresses the task directly. `any()` is useful when the condition is more complex. Use `.index()` when you actually need a position; it raises `ValueError` when absent.


```python
houses = ["Gryffindor", "Slytherin", "Ravenclaw", "Hufflepuff"]
print("Exact membership:", "Gryffindor" in houses)
print("Any long name:", any(len(house) > 10 for house in houses))
print("Position:", houses.index("Ravenclaw"))
try:
    houses.index("Durmstrang")
except ValueError:
    print("No such item; index() raises ValueError.")
```
