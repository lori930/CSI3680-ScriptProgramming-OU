### How to Run Bash Commands in Google Colab?
In Google Colab, you can run Bash commands directly by prefixing them with an exclamation mark (`!`) in a code cell. Example:
```
!pwd
!ls -l
!echo "Hello from Bash"
```
For multi-line shell scripts, you can use `%%bash` at the start of the cell:
```
%%bash
echo "Welcome to CSI3680"
echo "We do script programming here"
```

### A Magic Command `%%writefile`
One more thing, let's introduce a magic command that allows us to write a Shell/Python script to disk and then run it as if it were a standalone file.

Later on, we will learn how to use a *Here-Document* to create Bash scripts. Today, let's just make it simple: `%%writefile` magic command.


````python
%%writefile demo.sh
#!/bin/bash
name="world"
echo "Hello, ${name}"
````

Execute the above code block. Now you should see `demo.sh` showing up in `Files` from the left navigation bar.

Now we can run it with


````python
!bash demo.sh
````

You should see the output `Hello, world` 🤩

This magic command works for Python scripts too. For example:

````python
%%writefile demo.py
# Example script content
x = int(input("Enter a number: "))
print("Square:", x ** 2)
````


Execute the code block. Then we can run it with

````python
!python demo.py
````
