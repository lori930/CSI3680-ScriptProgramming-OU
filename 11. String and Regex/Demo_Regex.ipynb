**Putting character matching, quantifiers, anchors together.**

```python
import re

text = "Order IDs: AB123, CD456, XY78"

pattern = r"[A-Z]{2}\d{3}"

print(re.findall(pattern, text))
```

Run the code first, and explain:
1. `[A-Z]` matches what?
2. `{2}` repeats how many times?
3. `\d{3}` matches what?

What if the text contains  ABC1234 ? Will our current pattern match part of it? What if we set the pattern to be `pattern = r"\b[A-Z]{2}\d{3}\b"`?
