### Hint: stopping a script and reporting failure
You have seen `try/except ValueError` in class. This lab adds the exit-status argument to `sys.exit()`:

- After `import sys`, use `sys.exit(1)` to stop the script immediately and report failure to the shell. The `1` is an exit status, not a printed message.
- If the script finishes normally, its exit status is `0` (success). `sys.exit()` without an argument also reports `0`.
- Print the error message first. `print(message, file=sys.stderr)` sends it to standard error.

For example, this pattern rejects the wrong argument count before accessing arguments:

```python
import sys

if len(sys.argv) != 4:
    print("Usage: python3 calculator.py NUMBER1 OPERATOR NUMBER2", file=sys.stderr)
    sys.exit(1)

# This line is reached only when the argument count is correct.
```

Bash can inspect the exit status immediately afterward:

```bash
python3 calculator.py
printf 'Exit status: %s\n' "$?"
```

The usage message appears, followed by `Exit status: 1`. Reuse this print-then-exit pattern for other errors and for Part 3. A function's `return` sends a value back to its caller; `sys.exit(1)` ends the script and signals failure. Organize the remaining control flow yourself.
